# Subject-to-group spectral analysis

I aggregate subject estimates and match paired conditions by subject ID. These three generated subjects are **DEMONSTRATION ONLY**; their comparisons demonstrate the interface and have no physiological interpretation.

```python
from pathlib import Path
from IPython.display import Image, display
from eeg_workflow.cli import run_demo, run_group

repo = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
base = repo / "derivatives" / "notebook-group"
runs = []
for number, seed in enumerate((97, 98, 99), 1):
    subject = f"demo-{number:03d}"
    runs.append(run_demo(base / subject, seed=seed, subject_id=subject))
```

## Select a comparison

The feature table keeps ROI definitions, bands, units and epoch counts with each value. I choose metrics explicitly for this example. For an experimental study, define the comparisons and multiplicity handling in the analysis plan.

```python
group = base / "group"
result = run_group([run / "features.csv" for run in runs], group,
                   condition_a="A", condition_b="B",
                   curve_paths=[run / "psd_curves.csv" for run in runs],
                   metrics=["Frontal_Alpha_ln_mean_psd", "FAA"])
display(result)
display(Image(filename=str(group / "grand_average_psd.png")))
display(Image(filename=str(group / "FAA.png")))
```

The output reports complete pairs, excluded subject IDs and unadjusted p-values. Subject PSD curves receive equal weight in the grand average; individual epochs are not independent subjects. An SEM from three synthetic subjects is only an example of the calculation.
